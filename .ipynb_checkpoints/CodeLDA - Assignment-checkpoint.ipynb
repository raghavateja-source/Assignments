```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

np.random.seed(42)

# Generate data
n = 200

# Class 0
x0 = np.random.normal(loc=0, scale=1, size=n)
y0 = np.random.normal(loc=0, scale=5, size=n)

# Class 1
x1 = np.random.normal(loc=3, scale=1, size=n)
y1 = np.random.normal(loc=0, scale=5, size=n)

X = np.vstack((np.column_stack((x0, y0)),
               np.column_stack((x1, y1))))
y = np.array([0]*n + [1]*n)

# Plot original data
plt.scatter(x0, y0, label="Class 0", alpha=0.5)
plt.scatter(x1, y1, label="Class 1", alpha=0.5)
plt.legend()
plt.title("Original Data")
plt.show()

# PCA
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)

# LDA
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

# Plot projections
plt.hist(X_pca[y==0], alpha=0.5, label="Class 0")
plt.hist(X_pca[y==1], alpha=0.5, label="Class 1")
plt.title("PCA Projection")
plt.legend()
plt.show()

plt.hist(X_lda[y==0], alpha=0.5, label="Class 0")
plt.hist(X_lda[y==1], alpha=0.5, label="Class 1")
plt.title("LDA Projection")
plt.legend()
plt.show()
```

```python
Questions :

1.	Why does PCA choose the wrong direction here? 
2.	What information is PCA missing? 
3.	How does LDA use labels to fix this? 
4.	What would happen if class means were identical? 
```

```python
1.	Why does PCA choose the wrong direction here? 

#PCA chooses the direction with max variance

# Class 0
x0 = np.random.normal(loc=0, scale=1, size=n)
y0 = np.random.normal(loc=0, scale=5, size=n)

# Class 1
x1 = np.random.normal(loc=3, scale=1, size=n)
y1 = np.random.normal(loc=0, scale=5, size=n)


The classes differ in x 
(loc=0 vs loc=3).
The classes have the same y distribution (loc=0, scale=5).

The variance along each axis is approximately:

x variance ≈
=1
y variance ≈ 
=25

# As max variance is along Y - it has choosen Y direction 
```

```python
#2.	What information is PCA missing? 

pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)

# Plot projections
plt.hist(X_pca[y==0], alpha=0.5, label="Class 0")
plt.hist(X_pca[y==1], alpha=0.5, label="Class 1")
plt.title("PCA Projection")
plt.legend()
plt.show()
#only X value is passed on, It can recongnise y values have no knowledge of class 0, or classss 1



```

```python
#3.	How does LDA use labels to fix this? 


lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

# we have passed on both values X,y LDA recognises Class 0 & Class 1, also notices where is changing a lot in X, So it chooses X directions


```

```python
#4.	What would happen if class means were identical?

import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

np.random.seed(42)

# Generate data
n = 200

# Class 0
x0 = np.random.normal(loc=0, scale=1, size=n)
y0 = np.random.normal(loc=0, scale=5, size=n)

# Class 1
x1 = np.random.normal(loc=0, scale=1, size=n) #modified class
y1 = np.random.normal(loc=0, scale=5, size=n)

X = np.vstack((np.column_stack((x0, y0)),
               np.column_stack((x1, y1))))
y = np.array([0]*n + [1]*n)

# Plot original data
plt.scatter(x0, y0, label="Class 0", alpha=0.5)
plt.scatter(x1, y1, label="Class 1", alpha=0.5)
plt.legend()
plt.title("Original Data")
plt.show()

# PCA
pca = PCA(n_components=1)
X_pca = pca.fit_transform(X)

# LDA
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

# Plot projections
plt.hist(X_pca[y==0], alpha=0.5, label="Class 0")
plt.hist(X_pca[y==1], alpha=0.5, label="Class 1")
plt.title("PCA Projection")
plt.legend()
plt.show()


plt.hist(X_lda[y==0], alpha=0.5, label="Class 0")
plt.hist(X_lda[y==1], alpha=0.5, label="Class 1")
plt.title("LDA Projection")
plt.legend()
plt.show()
```
